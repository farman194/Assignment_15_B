```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\Farman Ali\Documents\semester 3\ICA\student_tea_snacks _1_.csv")

df
```

## Most prefered Tea Type

```python
tea_type_count = df.groupby("tea_type").count()

tea_type_count["student_id"].sort_values(ascending=False).head(1)
```

## Average NUmber of Cups by Gender

```python
gender_cups = df.groupby("gender")["cups_per_day"].mean()

gender_cups
```

## Snack Choices

```python
snack_counts = df["snack_choice"].value_counts()
print(snack_counts)
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(snack_counts.index.astype(str), snack_counts.values)
ax.set_title("Snack Choices Distribution")
ax.set_xlabel("Snack Choice")
ax.set_ylabel("Count")
ax.tick_params(axis="x", rotation=30)
ax.bar_label(bars, padding=3)
plt.tight_layout()
plt.show()
```

## compare tea preference by age group

```python
counts_by_age = pd.crosstab(df["age_group"], df["prefers_tea"])
print("Counts by Age Group:")
print(counts_by_age)
pct_by_age = pd.crosstab(df["age_group"], df["prefers_tea"], normalize="index") * 100
print("\nPercentage (%) Within Each Age Group:")
print(pct_by_age.round(1))
```
